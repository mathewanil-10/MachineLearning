```python
transactions = [
    ["Action", "Comedy", "Thriller"],
    ["Action", "Drama", "Thriller"],
    ["Comedy", "Thriller", "Romance"],
    ["Action", "Comedy", "Thriller"],
    ["Action", "Comedy", "Drama"]
]
```

```python
min_support = 3
```

```python
count = {}
```

```python
for transaction in transactions:
    for item in transaction:
        if item in count:
            count[item] = count[item] + 1
        else:
            count[item] = 1
```

```python
print("Item Counts")
print(count)
```

```python
frequent = {}
```

```python
for item in count:
    if count[item] >= min_support:
        frequent[item] = count[item]
```

```python
print("\nFrequent Items")
print(frequent)
```

```python
new_transactions = []
```

```python
for transaction in transactions:

    temp = []

    for item in transaction:
        if item in frequent:
            temp.append(item)

    new_transactions.append(temp)
```

```python
print("\nTransactions after removing infrequent items")

for t in new_transactions:
    print(t)
```

```python
tree = {}
```

```python
for transaction in new_transactions:

    current = tree

    for item in transaction:

        if item not in current:
            current[item] = {"count":1,"child":{}}

        else:
            current[item]["count"] += 1

        current = current[item]["child"]
```

```python
def show(tree, space=""):

    for item in tree:
        print(space + item + " (" + str(tree[item]["count"]) + ")")
        show(tree[item]["child"], space + "   ")
```

```python
print("\nFP Tree")
show(tree)
```

**INTERPRETATION**

The FP-Tree identifies the most frequently watched movie genres.
Action, Comedy, and Thriller are the most common genres.
These patterns help improve the recommendation system.
This provides users with better movie suggestions.
