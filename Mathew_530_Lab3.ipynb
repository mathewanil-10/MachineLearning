```python
from itertools import combinations
```

```python
transactions = [
    ["Paracetamol", "Vitamin C", "Cough Syrup"],
    ["Paracetamol", "Antibiotic", "Vitamin C"],
    ["Antibiotic", "Vitamin C", "Pain Reliever"],
    ["Paracetamol", "Vitamin C", "Pain Reliever"],
    ["Paracetamol", "Antibiotic", "Vitamin C"],
    ["Vitamin C", "Cough Syrup"]
]

min_support = 3
```

```python
C1 = {}
```

```python
for transaction in transactions:
    for item in transaction:
        if item in C1:
            C1[item] += 1
        else:
            C1[item] = 1
```

```python
print("C1")
print(C1)
```

```python
L1 = {}
```

```python
for item in C1:
    if C1[item] >= min_support:
        L1[item] = C1[item]
```

```python
print("\nL1")
print(L1)
```

```python
C2 = {}
```

```python
for transaction in transactions:

    temp = []

    for item in transaction:
        if item in L1:
            temp.append(item)

    pairs = combinations(sorted(temp), 2)

    for pair in pairs:
        if pair in C2:
            C2[pair] += 1
        else:
            C2[pair] = 1
```

```python
print("\nC2")
print(C2)
```

```python
L2 = {}
```

```python
for pair in C2:
    if C2[pair] >= min_support:
        L2[pair] = C2[pair]
```

```python
print("\nL2")
print(L2)
```

```python
C3 = {}
```

```python
for transaction in transactions:

    temp = []

    for item in transaction:
        if item in L1:
            temp.append(item)

    triples = combinations(sorted(temp), 3)

    for triple in triples:
        if triple in C3:
            C3[triple] += 1
        else:
            C3[triple] = 1
```

```python
print("\nC3")
print(C3)
```

```python
L3 = {}
```

```python
for triple in C3:
    if C3[triple] >= min_support:
        L3[triple] = C3[triple]
```

```python
print("\nL3")
print(L3)
```

**INTERPRETATION**

The Apriori algorithm was applied to the hospital prescription records to identify medicines that are frequently prescribed together. By setting the minimum support count to 3, only medicines and medicine combinations that appeared at least three times were considered frequent.

The results showed that **Vitamin C**, **Paracetamol**, and **Antibiotic** were the most frequently prescribed medicines. The combinations **Paracetamol and Vitamin C** and **Antibiotic and Vitamin C** were also frequently found in the prescriptions. The three-medicine combination **Paracetamol, Antibiotic, and Vitamin C** satisfied the minimum support count and was identified as a frequent itemset.

Using the Apriori pruning principle, medicines and combinations with low support were removed from further analysis. Based on these results, the hospital can stock these medicines together and prepare medicine kits more efficiently. This helps improve inventory management, reduces shortages, and ensures that commonly prescribed medicines are readily available for patients.
