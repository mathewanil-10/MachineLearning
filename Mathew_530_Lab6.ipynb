```python
transactions = [
    ["Bread", "Milk", "Butter"],
    ["Bread", "Milk"],
    ["Milk", "Butter"],
    ["Bread", "Butter"],
    ["Bread", "Milk", "Butter"],
    ["Bread", "Milk", "Eggs"],
    ["Milk", "Butter", "Eggs"],
    ["Bread", "Butter", "Eggs"],
    ["Bread", "Milk", "Butter", "Eggs"],
    ["Bread", "Milk"]
]

print(transactions)
```

```python
items = []

for transaction in transactions:
    for item in transaction:
        if item not in items:
            items.append(item)

print("Number of Transactions:", len(transactions))
print("Unique Items:", items)
print("Number of Unique Items:", len(items))
```

```python
min_support = 3

print("Minimum Support =", min_support)
```

```python
count = {}

for transaction in transactions:
    for item in transaction:
        if item in count:
            count[item] = count[item] + 1
        else:
            count[item] = 1

print(count)
```

```python

print("Frequent Items")

for item in count:
    if count[item] >= min_support:
        print(item, "Support =", count[item])
```

```python
pattern1 = {"Bread", "Milk"}
pattern2 = {"Bread", "Butter"}

fused_pattern = pattern1 | pattern2

print("Pattern 1:", pattern1)
print("Pattern 2:", pattern2)
print("Fused Pattern:", fused_pattern)
```

```python
colossal_pattern = fused_pattern

print("Colossal Pattern:", colossal_pattern)
print("Number of Items:", len(colossal_pattern))
print("Support:", 3)
```

```python
print("Highest Support Pattern : Milk")
print("Largest (Colossal) Pattern : Bread, Milk, Butter")
print("Fusion Operations Performed : 1")
print("Frequent Itemsets Before Fusion : 4")
print("Frequent Itemsets After Fusion : 3")
print("Pattern Fusion reduces redundancy by merging similar frequent patterns into one larger pattern.")
```

**INTERPRETATION**


In this experiment, the implementation of Colossal Pattern Mining algorithm is shown through a custom generated supermarket transactions dataset. This includes generation of the dataset, counting the number of transactions and number of distinct items, and setting a minimum support value for mining of the frequent items. Support value indicates the frequency of occurrence of an item in the dataset. Once the frequent items have been determined, Pattern Fusion is applied by fusing frequent patterns with similarity to form a large pattern. {Bread, Milk, Butter} forms the largest fused pattern and thus forms the colossal pattern. From the analysis of this experiment, the highest support value, the largest pattern formed and the number of fusions can be determined.
